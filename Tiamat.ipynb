<h2>Introdução</h2>

O Enem (Exame Nacional do Ensino Médio), criado em 1998, é uma prova realizada anualmente no Brasil, com o objetivo de avaliar o desempenho dos estudantes ao fim da educação básica. A partir de 2009, passou a ser utilizado como forma de acesso ao ensino superior [1].

As provas envolvem quatro áreas do conhecimento: linguagens, códigos e suas tecnologias; ciências humanas e suas tecnologias; ciências da natureza e suas tecnologias; e matemática e suas tecnologias, o que totaliza 180 questões objetivas. Além disso, o exame inclui ainda uma redação, composta pela produção de um texto dissertativo-argumentativo a partir de uma situação-problema [1].

As notas das questões objetivas do Enem são dadas por meio do TRI (Teoria de Resposta ao Item), um modelo matemático que analisa o padrão de respostas do candidato em relação ao conjunto de questões que formam o exame, priorizando o desempenho que for mais coerente. O TRI qualifica as questões com base em três parâmetros: grau de dificuldade, discriminação e acerto ao acaso (chute) [2].

Com as notas do TRI para cada uma das quatro áreas do conhecimento, soma-se a nota atingida pelo candidato na redação e, dividindo essa soma por 5, obtém-se a média total com redação. 

Com o intuito de rankear as escolas por todo o Brasil, determinam-se as médias totais, com e sem redação, de cada escola que teve alunos participantes no Enem de cada ano. 

<--- É justamente a média total com redação que vamos usar nesse trabalho para treinar e testar um modelo KNN. Em outras palavras, para essa atividade da disciplina de Aprendizado de Máquina, esse será nosso *target* (o alvo). Os atributos usados serão as posições municipal, estadual e nacional das escolas e a média total sem redação. --->

<h2>Desenvolvimento</h2>

Inicialmente vou importar os módulos e funções que serão necessárias para meu trabalho.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.metrics import mean_squared_error, root_mean_squared_error
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    MinMaxScaler,
    OneHotEncoder,
    OrdinalEncoder,
    StandardScaler,
)
```

Agora, vamos transformar os dados do ranking, que estão em uma planilha Excel, para o formato de Data Frame do Python, por meio do módulo pandas. Ademais, vamos separar as colunas da planilha que serão usadas como atributos (que chamarei de X) e a que será nosso *target* (que nomearei y).

```python
NOME_ARQUIVO = "Enem_Acompanhamento_de_escolas - Copia.xlsx"
ATRIBUTOS = ["Cidade", 
             "Estado", 
             "Posição municipal (com redação)", 
             "Posição municipal (objetivas)",
             "Posição estadual (com redação)",
             "Posição estadual (objetivas)",
             "Posição total (com redação)",
             "Posição total (objetivas)",
             "Média nacional (objetivas)",         
            ]
TARGET = ["Média nacional (com redação)"]

df = pd.read_excel(NOME_ARQUIVO)
df = df.dropna()

X = df[ATRIBUTOS]
y = df[TARGET]
```

Abaixo, mostro os atributos (X) e o target (y).

```python
X
```

```python
y
```

```python
SEED = 42

indices_treino, indices_teste = train_test_split(
    df.index, test_size=0.1, random_state=SEED
)

df_treino = df.loc[indices_treino]
df_teste = df.loc[indices_teste]

X_treino = df_treino.reindex(ATRIBUTOS, axis=1)
y_treino = df_treino.reindex(TARGET, axis=1)

X_teste = df_teste.reindex(ATRIBUTOS, axis=1)
y_teste = df_teste.reindex(TARGET, axis=1)
```

```python
X_treino
```

```python
COLUNAS_CATEGORICAS = ["Cidade", "Estado"]

ohe = OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore")

ohe_treino = ohe.fit_transform(X_treino[COLUNAS_CATEGORICAS])
ohe_teste = ohe.transform(X_teste[COLUNAS_CATEGORICAS])

ohe_colunas = ohe.get_feature_names_out(COLUNAS_CATEGORICAS)
df_ohe_treino = pd.DataFrame(
    ohe_treino, columns=ohe_colunas, index=X_treino.index
)
df_ohe_teste = pd.DataFrame(ohe_teste, columns=ohe_colunas, index=X_teste.index)

COLUNAS_NUMERICAS = []

for col in ATRIBUTOS:
    if col not in COLUNAS_CATEGORICAS:
        COLUNAS_NUMERICAS.append(col)

X_treino_encoded = pd.concat(
    [X_treino[COLUNAS_NUMERICAS], df_ohe_treino], axis=1
)
X_teste_encoded = pd.concat([X_teste[COLUNAS_NUMERICAS], df_ohe_teste], axis=1)

NUM_VIZINHOS = 3
modelo_knn = KNeighborsRegressor(n_neighbors=NUM_VIZINHOS)
modelo_knn.fit(X_treino_encoded, y_treino.values.ravel())
```

```python
X_verdadeiro = X_teste_encoded
y_verdadeiro = y_teste

y_previsto = modelo_knn.predict(X_verdadeiro)

RMSE = root_mean_squared_error(y_verdadeiro, y_previsto)

print(f"O RMSE do modelo foi de{RMSE: .2f} unidades de y.")
```

```python
# só para eu testar!

dado_exemplo = X_teste_encoded.iloc[[20]]
valor_real = y_teste.iloc[20].values[0]

previsao = modelo_knn.predict(dado_exemplo)[0]

print(f"Valor REAL: {valor_real:.2f}")
print(f"Valor PREVISTO pelo KNN: {previsao:.2f}")
print(f"Diferença (Erro): {abs(valor_real - previsao):.2f}")
```

```python
y_real = y_verdadeiro.values.ravel()
y_prev = y_previsto.ravel()

min_val = min(y_real.min(), y_prev.min())
max_val = max(y_real.max(), y_prev.max())

intervalo = [min_val * 0.95, max_val * 1.05]
```

```python
eixo = sns.scatterplot(x=y_real, y=y_prev, label="Dados", alpha=0.7, s=40)
eixo.plot(intervalo, intervalo, "r--", label="Identidade", linewidth=1.5)

eixo.set(
    xlabel="$y$ verdadeiro",
    ylabel="$y$ previsto",
    xlim=intervalo,
    ylim=intervalo,
    aspect="equal",
)
eixo.legend();
```

```python
combinacoes = [
    {"n_neighbors": 1, "weights": "distance", "p": 2},
    {"n_neighbors": 3, "weights": "distance", "p": 2},
    {"n_neighbors": 5, "weights": "distance", "p": 2},
    {"n_neighbors": 7, "weights": "distance", "p": 2},
    {"n_neighbors": 10, "weights": "distance", "p": 2},
    {"n_neighbors": 15, "weights": "distance", "p": 2},
    {"n_neighbors": 20, "weights": "distance", "p": 2},
    {"n_neighbors": 3, "weights": "distance", "p": 1},
    {"n_neighbors": 5, "weights": "distance", "p": 1},
    {"n_neighbors": 7, "weights": "distance", "p": 1},
]

resultados = []

for i, params in enumerate(combinacoes, start=1):
    knn = KNeighborsRegressor(**params, algorithm="kd_tree", n_jobs=-1)

    knn.fit(X_treino_encoded, y_treino.values.ravel())

    y_prev = knn.predict(X_teste_encoded)
    rmse = root_mean_squared_error(y_teste, y_prev)

    resultados.append(
        {
            "Conjunto": f"Hiperparâmetro {i}",
            "K (n_neighbors)": params["n_neighbors"],
            "Peso (weights)": params["weights"],
            "Métrica (p)": "Manhattan" if params["p"] == 1 else "Euclidiana",
            "RMSE": rmse,
        }
    )

df_resultados = pd.DataFrame(resultados).sort_values(by="RMSE")
print(df_resultados.to_string(index=False))

# forma manual de realizar as combinações dos hiperparâmetros
```

```python
knn = KNeighborsRegressor(algorithm="kd_tree")

param_grid = {
    "n_neighbors": [1, 3, 5, 7, 10],  
    "weights": ["distance"],  
    "p": [1, 2],
}  

grid_search = GridSearchCV(
    estimator=knn,
    param_grid=param_grid,
    scoring="neg_root_mean_squared_error", 
    cv=3, 
    n_jobs=-1,  
)

grid_search.fit(X_treino_encoded, y_treino.values.ravel())

df_grid = pd.DataFrame(grid_search.cv_results_)
df_grid["RMSE"] = -df_grid["mean_test_score"] 

tabela_relatorio = df_grid[
    [
        "param_n_neighbors",
        "param_weights",
        "param_p",
        "RMSE",
        "rank_test_score",
    ]
].sort_values(by="rank_test_score")

tabela_relatorio.columns = [
    "K (n_neighbors)",
    "Peso",
    "Métrica (p)",
    "RMSE",
    "Ranking",
]
print(tabela_relatorio.to_string(index=False))

melhor_modelo = grid_search.best_estimator_
y_prev = melhor_modelo.predict(X_teste_encoded)
rmse_teste = root_mean_squared_error(y_teste, y_prev)

print(f"\nMelhor combinação encontrada: {grid_search.best_params_}")
print(f"RMSE final no conjunto de teste: {rmse_teste:.4f}")
```
